```python
# ============================================================
# MICROVISION AI
# Complete CNN Analysis for Microplastic Classification
# Google Colab - Run Everything in ONE CELL
# ============================================================

import os
import zipfile
import shutil
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

IMG_SIZE = (224, 224)
BATCH_SIZE = 16
EPOCHS = 15

WORK_DIR = Path("/content/MicroVision_AI")
EXTRACT_DIR = WORK_DIR / "dataset"
MODEL_DIR = WORK_DIR / "models"

WORK_DIR.mkdir(parents=True, exist_ok=True)
EXTRACT_DIR.mkdir(parents=True, exist_ok=True)
MODEL_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 80)
print("MICROVISION AI - MICROPLASTIC CNN CLASSIFICATION")
print("=" * 80)

# ============================================================
# 2. UPLOAD DATASET ZIP
# ============================================================

try:
    from google.colab import files

    print("\nUpload dataset_microplastics-master.zip")
    uploaded = files.upload()

    zip_files = [
        filename for filename in uploaded.keys()
        if filename.lower().endswith(".zip")
    ]

    if not zip_files:
        raise FileNotFoundError("No ZIP file was uploaded.")

    zip_path = Path("/content") / zip_files[0]

except Exception as error:
    print("Colab upload was not available.")
    print("Trying to use the already-mounted uploaded ZIP file.")

    possible_paths = [
        Path("/mnt/data/dataset_microplastics-master.zip"),
        Path("/content/dataset_microplastics-master.zip")
    ]

    zip_path = None

    for candidate in possible_paths:
        if candidate.exists():
            zip_path = candidate
            break

    if zip_path is None:
        raise FileNotFoundError(
            "Dataset ZIP was not found. Please upload dataset_microplastics-master.zip."
        )

print(f"\nZIP file found: {zip_path}")

# ============================================================
# 3. EXTRACT DATASET
# ============================================================

if EXTRACT_DIR.exists():
    shutil.rmtree(EXTRACT_DIR)

EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(EXTRACT_DIR)

print("Dataset extracted successfully.")

# ============================================================
# 4. FIND CLASS DIRECTORIES
# ============================================================

image_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".tif",
    ".tiff"
}

all_images = []

for image_path in EXTRACT_DIR.rglob("*"):
    if image_path.is_file() and image_path.suffix.lower() in image_extensions:
        all_images.append(image_path)

if len(all_images) == 0:
    raise RuntimeError("No image files were found inside the ZIP dataset.")

# The uploaded dataset normally contains:
# dataset_microplastics-master/micro_plastic/<class_name>/image.jpg

class_directories = {}

for image_path in all_images:
    relative_parts = image_path.relative_to(EXTRACT_DIR).parts

    if len(relative_parts) < 2:
        continue

    # Find the folder directly containing the image
    class_name = image_path.parent.name

    if class_name.lower() in {
        "micro_plastic",
        "microplastic",
        "dataset",
        "images"
    }:
        continue

    class_directories.setdefault(class_name, []).append(image_path)

# Remove folders with too few images
class_directories = {
    class_name: paths
    for class_name, paths in class_directories.items()
    if len(paths) >= 2
}

if len(class_directories) < 2:
    raise RuntimeError(
        "At least two classes with images are required for classification."
    )

print("\nDetected classes:")
for class_name, paths in sorted(class_directories.items()):
    print(f"{class_name:25s} : {len(paths):4d} images")

# ============================================================
# 5. BUILD DATAFRAME
# ============================================================

records = []

for class_name, paths in sorted(class_directories.items()):
    for image_path in paths:
        records.append({
            "filepath": str(image_path),
            "label": class_name
        })

df = pd.DataFrame(records)

class_names = sorted(df["label"].unique())
class_to_index = {
    class_name: index
    for index, class_name in enumerate(class_names)
}

df["label_id"] = df["label"].map(class_to_index)

print("\nTotal images:", len(df))
print("Number of classes:", len(class_names))
print("Class names:", class_names)

# ============================================================
# 6. VISUALIZE CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(12, 5))
sns.countplot(
    data=df,
    x="label",
    order=df["label"].value_counts().index
)
plt.title("Microplastic Dataset Class Distribution")
plt.xlabel("Microplastic Class")
plt.ylabel("Number of Images")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ============================================================
# 7. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# First split: 80% training, 20% temporary
train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    stratify=df["label_id"],
    random_state=SEED
)

# Second split: temporary into validation and test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label_id"],
    random_state=SEED
)

print("\nDataset split:")
print("Training images:  ", len(train_df))
print("Validation images:", len(val_df))
print("Testing images:   ", len(test_df))

# ============================================================
# 8. IMAGE DATA GENERATORS
# ============================================================

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
    layers.RandomContrast(0.15)
], name="data_augmentation")

def load_image(filepath, label):
    image = tf.io.read_file(filepath)
    image = tf.image.decode_image(
        image,
        channels=3,
        expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)
    image = preprocess_input(image)
    return image, label

def create_dataset(dataframe, training=False):
    filepaths = dataframe["filepath"].values
    labels = dataframe["label_id"].values

    dataset = tf.data.Dataset.from_tensor_slices(
        (filepaths, labels)
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=len(dataframe),
            seed=SEED
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if training:
        dataset = dataset.map(
            lambda image, label: (
                data_augmentation(image, training=True),
                label
            ),
            num_parallel_calls=tf.data.AUTOTUNE
        )

    dataset = dataset.batch(BATCH_SIZE)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset

train_ds = create_dataset(train_df, training=True)
val_ds = create_dataset(val_df, training=False)
test_ds = create_dataset(test_df, training=False)

# ============================================================
# 9. DISPLAY SAMPLE IMAGES
# ============================================================

plt.figure(figsize=(14, 8))

sample_df = train_df.sample(
    min(12, len(train_df)),
    random_state=SEED
)

for index, (_, row) in enumerate(sample_df.iterrows()):
    image = tf.keras.utils.load_img(
        row["filepath"],
        target_size=IMG_SIZE
    )

    image_array = tf.keras.utils.img_to_array(image) / 255.0

    plt.subplot(3, 4, index + 1)
    plt.imshow(image_array)
    plt.title(row["label"])
    plt.axis("off")

plt.suptitle("Sample Microplastic Images", fontsize=16)
plt.tight_layout()
plt.show()

# ============================================================
# 10. BUILD CNN MODEL
# ============================================================

num_classes = len(class_names)

base_model = MobileNetV2(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

# Freeze the pretrained CNN initially
base_model.trainable = False

inputs = layers.Input(
    shape=(224, 224, 3),
    name="microplastic_image"
)

x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.35)(x)
x = layers.Dense(
    128,
    activation="relu",
    name="feature_layer"
)(x)
x = layers.Dropout(0.25)(x)

outputs = layers.Dense(
    num_classes,
    activation="softmax",
    name="prediction"
)(x)

model = models.Model(
    inputs=inputs,
    outputs=outputs,
    name="MicroVision_MobileNetV2_CNN"
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# ============================================================
# 11. TRAIN CNN
# ============================================================

checkpoint_path = MODEL_DIR / "microvision_best.keras"

callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-7,
        verbose=1
    ),
    ModelCheckpoint(
        filepath=str(checkpoint_path),
        monitor="val_accuracy",
        save_best_only=True,
        verbose=1
    )
]

print("\nStarting CNN training...")

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

# ============================================================
# 12. OPTIONAL FINE-TUNING
# ============================================================

print("\nStarting light fine-tuning...")

base_model.trainable = True

# Freeze most layers and fine-tune only the final layers
for layer in base_model.layers[:-30]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

fine_tune_history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5,
    callbacks=callbacks,
    verbose=1
)

# ============================================================
# 13. TRAINING CURVES
# ============================================================

def combine_history_values(history_one, history_two, key):
    values_one = history_one.history.get(key, [])
    values_two = history_two.history.get(key, [])
    return values_one + values_two

train_accuracy = combine_history_values(
    history,
    fine_tune_history,
    "accuracy"
)

val_accuracy = combine_history_values(
    history,
    fine_tune_history,
    "val_accuracy"
)

train_loss = combine_history_values(
    history,
    fine_tune_history,
    "loss"
)

val_loss = combine_history_values(
    history,
    fine_tune_history,
    "val_loss"
)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(train_accuracy, label="Training Accuracy")
plt.plot(val_accuracy, label="Validation Accuracy")
plt.title("CNN Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(train_loss, label="Training Loss")
plt.plot(val_loss, label="Validation Loss")
plt.title("CNN Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================
# 14. EVALUATE MODEL
# ============================================================

test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

print("\nFinal Test Results")
print("-" * 50)
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")

# ============================================================
# 15. PREDICTIONS
# ============================================================

y_true = []
y_pred = []
y_probability = []

for images, labels in test_ds:
    probabilities = model.predict(
        images,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    y_true.extend(labels.numpy())
    y_pred.extend(predictions)
    y_probability.extend(probabilities)

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_probability = np.array(y_probability)

# ============================================================
# 16. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report")
print("=" * 80)

print(
    classification_report(
        y_true,
        y_pred,
        labels=list(range(num_classes)),
        target_names=class_names,
        zero_division=0
    )
)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print(f"Weighted Precision: {precision:.4f}")
print(f"Weighted Recall:    {recall:.4f}")
print(f"Weighted F1-Score:  {f1:.4f}")

# ============================================================
# 17. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(num_classes))
)

plt.figure(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Microplastic Classification Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 18. CLASS-WISE ACCURACY
# ============================================================

class_accuracy = {}

for class_id, class_name in enumerate(class_names):
    class_indices = y_true == class_id

    if np.sum(class_indices) > 0:
        class_accuracy[class_name] = np.mean(
            y_pred[class_indices] == y_true[class_indices]
        )
    else:
        class_accuracy[class_name] = 0

plt.figure(figsize=(12, 5))
plt.bar(
    class_accuracy.keys(),
    class_accuracy.values()
)

plt.title("Class-Wise Accuracy")
plt.xlabel("Class")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================
# 19. GRAD-CAM EXPLAINABILITY
# ============================================================

def make_gradcam_heatmap(image_array, trained_model, last_conv_layer_name):
    grad_model = tf.keras.models.Model(
        inputs=trained_model.inputs,
        outputs=[
            trained_model.get_layer(last_conv_layer_name).output,
            trained_model.output
        ]
    )

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(image_array)
        predicted_class = tf.argmax(
            predictions[0]
        )
        class_channel = predictions[:, predicted_class]

    gradients = tape.gradient(
        class_channel,
        conv_outputs
    )

    pooled_gradients = tf.reduce_mean(
        gradients,
        axis=(0, 1, 2)
    )

    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_gradients[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(heatmap, 0)
    heatmap = heatmap / (
        tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
    )

    return heatmap.numpy(), int(predicted_class)


# Find the last convolutional layer automatically
last_conv_layer = None

for layer in reversed(model.layers):
    try:
        output_shape = layer.output.shape

        if len(output_shape) == 4:
            last_conv_layer = layer.name
            break
    except Exception:
        continue

print("\nGrad-CAM layer:", last_conv_layer)

# ============================================================
# 20. DISPLAY GRAD-CAM RESULTS
# ============================================================

gradcam_samples = test_df.sample(
    min(6, len(test_df)),
    random_state=SEED
)

plt.figure(figsize=(15, 10))

for index, (_, row) in enumerate(gradcam_samples.iterrows()):
    original_image = tf.keras.utils.load_img(
        row["filepath"],
        target_size=IMG_SIZE
    )

    image_array = tf.keras.utils.img_to_array(
        original_image
    )

    normalized_image = preprocess_input(
        image_array.copy()
    )

    batch_image = np.expand_dims(
        normalized_image,
        axis=0
    )

    probabilities = model.predict(
        batch_image,
        verbose=0
    )[0]

    predicted_id = int(np.argmax(probabilities))
    predicted_label = class_names[predicted_id]
    confidence = float(probabilities[predicted_id])

    heatmap, _ = make_gradcam_heatmap(
        batch_image,
        model,
        last_conv_layer
    )

    plt.subplot(2, 3, index + 1)
    plt.imshow(original_image)
    plt.imshow(
        heatmap,
        cmap="jet",
        alpha=0.45,
        extent=(0, 224, 224, 0)
    )

    plt.title(
        f"Actual: {row['label']}\n"
        f"Predicted: {predicted_label}\n"
        f"Confidence: {confidence:.2%}"
    )

    plt.axis("off")

plt.suptitle(
    "Grad-CAM Explainability for Microplastic Classification",
    fontsize=16
)

plt.tight_layout()
plt.show()

# ============================================================
# 21. SINGLE IMAGE PREDICTION FUNCTION
# ============================================================

def predict_microplastic(image_path):
    image = tf.keras.utils.load_img(
        image_path,
        target_size=IMG_SIZE
    )

    image_array = tf.keras.utils.img_to_array(image)
    image_array = preprocess_input(image_array)
    image_batch = np.expand_dims(image_array, axis=0)

    probabilities = model.predict(
        image_batch,
        verbose=0
    )[0]

    predicted_id = int(np.argmax(probabilities))
    predicted_label = class_names[predicted_id]
    confidence = float(probabilities[predicted_id])

    plt.figure(figsize=(6, 6))
    plt.imshow(
        tf.keras.utils.load_img(
            image_path,
            target_size=IMG_SIZE
        )
    )
    plt.title(
        f"Prediction: {predicted_label}\n"
        f"Confidence: {confidence:.2%}"
    )
    plt.axis("off")
    plt.show()

    probability_table = pd.DataFrame({
        "Class": class_names,
        "Probability": probabilities
    }).sort_values(
        "Probability",
        ascending=False
    )

    display(probability_table)

    return predicted_label, confidence, probability_table

# Test prediction using one image from the test set
example_test_image = test_df.iloc[0]["filepath"]

print("\nExample Prediction")
print("=" * 50)
print("Image:", example_test_image)

predict_microplastic(example_test_image)

# ============================================================
# 22. SAVE MODEL AND METADATA
# ============================================================

final_model_path = MODEL_DIR / "MicroVision_Final_CNN.keras"
class_names_path = MODEL_DIR / "class_names.json"
metrics_path = MODEL_DIR / "evaluation_metrics.json"

model.save(final_model_path)

with open(class_names_path, "w") as file:
    import json
    json.dump(class_names, file, indent=4)

evaluation_metrics = {
    "test_loss": float(test_loss),
    "test_accuracy": float(test_accuracy),
    "weighted_precision": float(precision),
    "weighted_recall": float(recall),
    "weighted_f1": float(f1),
    "number_of_classes": int(num_classes),
    "class_names": class_names,
    "training_images": int(len(train_df)),
    "validation_images": int(len(val_df)),
    "testing_images": int(len(test_df))
}

with open(metrics_path, "w") as file:
    json.dump(evaluation_metrics, file, indent=4)

print("\n" + "=" * 80)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)
print("Saved model:       ", final_model_path)
print("Saved class names: ", class_names_path)
print("Saved metrics:     ", metrics_path)
print("Project directory: ", WORK_DIR)
print("=" * 80)
```

```python
# ============================================================
# FIXED GRAD-CAM CODE FOR MICROVISION CNN
# Run this after model training
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# 1. Get the MobileNetV2 base model
# ------------------------------------------------------------

base_model = None

for layer in model.layers:
    if isinstance(layer, tf.keras.Model):
        base_model = layer
        break

if base_model is None:
    raise RuntimeError("MobileNetV2 base model was not found.")

print("Base model:", base_model.name)

# ------------------------------------------------------------
# 2. Find the last real convolutional layer
# ------------------------------------------------------------

last_conv_layer = None

for layer in reversed(base_model.layers):
    if isinstance(
        layer,
        (
            tf.keras.layers.Conv2D,
            tf.keras.layers.DepthwiseConv2D,
            tf.keras.layers.SeparableConv2D
        )
    ):
        last_conv_layer = layer
        break

if last_conv_layer is None:
    raise RuntimeError("No convolutional layer was found.")

print("Correct Grad-CAM layer:", last_conv_layer.name)

# ------------------------------------------------------------
# 3. Create feature extractor
# ------------------------------------------------------------

feature_extractor = tf.keras.models.Model(
    inputs=base_model.input,
    outputs=[
        last_conv_layer.output,
        base_model.output
    ]
)

# ------------------------------------------------------------
# 4. Store classification head layers
# ------------------------------------------------------------

base_model_index = model.layers.index(base_model)
classification_head = model.layers[base_model_index + 1:]

print("\nClassification head layers:")
for layer in classification_head:
    print(" -", layer.name)

# ------------------------------------------------------------
# 5. Grad-CAM function
# ------------------------------------------------------------

def make_gradcam_heatmap(image_batch):
    """
    Creates a Grad-CAM heatmap for the trained MicroVision model.
    """

    with tf.GradientTape() as tape:

        conv_outputs, base_features = feature_extractor(
            image_batch,
            training=False
        )

        x = base_features

        # Pass features through the classification head
        for layer in classification_head:
            if isinstance(
                layer,
                (
                    tf.keras.layers.Dropout,
                    tf.keras.layers.BatchNormalization
                )
            ):
                x = layer(x, training=False)
            else:
                x = layer(x)

        predictions = x

        predicted_class = tf.argmax(
            predictions[0]
        )

        class_score = predictions[:, predicted_class]

    gradients = tape.gradient(
        class_score,
        conv_outputs
    )

    # Average gradients across spatial dimensions
    pooled_gradients = tf.reduce_mean(
        gradients,
        axis=(0, 1, 2)
    )

    conv_outputs = conv_outputs[0]

    # Weighted combination of feature maps
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_gradients,
        axis=-1
    )

    # Keep only positive activations
    heatmap = tf.maximum(
        heatmap,
        0
    )

    # Normalize between 0 and 1
    heatmap = heatmap / (
        tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
    )

    return heatmap.numpy(), int(predicted_class), predictions.numpy()[0]

# ------------------------------------------------------------
# 6. Generate Grad-CAM visualizations
# ------------------------------------------------------------

gradcam_samples = test_df.sample(
    min(6, len(test_df)),
    random_state=42
)

plt.figure(figsize=(15, 10))

for index, (_, row) in enumerate(
    gradcam_samples.iterrows()
):

    # Load original image
    original_image = tf.keras.utils.load_img(
        row["filepath"],
        target_size=IMG_SIZE
    )

    original_array = tf.keras.utils.img_to_array(
        original_image
    )

    # Preprocess for MobileNetV2
    processed_array = tf.keras.applications.mobilenet_v2.preprocess_input(
        original_array.copy()
    )

    image_batch = np.expand_dims(
        processed_array,
        axis=0
    )

    # Prediction
    probabilities = model.predict(
        image_batch,
        verbose=0
    )[0]

    predicted_id = int(
        np.argmax(probabilities)
    )

    predicted_label = class_names[predicted_id]
    confidence = float(probabilities[predicted_id])

    # Grad-CAM
    heatmap, _, _ = make_gradcam_heatmap(
        image_batch
    )

    # Resize heatmap to image size
    heatmap_resized = tf.image.resize(
        heatmap[..., np.newaxis],
        IMG_SIZE
    ).numpy().squeeze()

    # Display
    plt.subplot(2, 3, index + 1)

    plt.imshow(
        original_image
    )

    plt.imshow(
        heatmap_resized,
        cmap="jet",
        alpha=0.45
    )

    plt.title(
        f"Actual: {row['label']}\n"
        f"Predicted: {predicted_label}\n"
        f"Confidence: {confidence:.2%}"
    )

    plt.axis("off")

plt.suptitle(
    "MicroVision AI - Correct Grad-CAM Explainability",
    fontsize=16
)

plt.tight_layout()
plt.show()

print("\nGrad-CAM completed successfully.")
```
